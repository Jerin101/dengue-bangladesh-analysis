```python
print("Shape:", df.shape)
print()
df.info()
print()
print("Missing:")
print(df.isnull().sum())
print()
print("Duplicate:", df.duplicated().sum())
```

```python
df.columns = df.columns.str.strip().str.lower()
print(df.columns.tolist())
```

```python
df[["division", "year"]] = df[["division", "year"]].ffill()
df["year"] = df["year"].astype(int)

print(df["division"].value_counts())
print(df["year"].value_counts().sort_index())
print("Duplicate:", df.duplicated().sum())
print(df.isnull().sum().sum())
```

```python
print(df[["temperature", "pressure"]].head())
print(df["temperature"].dtype)
```

```python
for c in ["temperature", "pressure"]:
    df[c] = df[c].str.extract(r"(\d+\.?\d*)")[0].astype(float)

print(df[["temperature", "pressure"]].head())
print(df[["temperature", "pressure"]].dtypes)
print(df[["temperature", "pressure"]].describe())
```

```python
print(sorted(df["month"].unique()))
```

```python
df["date"] = pd.to_datetime(
    df["year"].astype(str) + "-" + df["month"] + "-01",
    format="%Y-%B-%d"
)
df[["year", "month", "date"]].head(3)
```

```python
df["death_rate"] = (df["death"] / df["infected"]).where(df["infected"] > 0, 0)
df["death_rate"].describe()
```
